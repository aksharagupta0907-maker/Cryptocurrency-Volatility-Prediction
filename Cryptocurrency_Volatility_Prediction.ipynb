**Data Preprocessing Required**

To fulfill the Data Preprocessing Required steps for financial, trading, or time-series data:

1. Handle Missing Values & Consistency

• Identify gaps: Check for NaN values from feed dropouts or halted trading hours.

• Impute or drop: Use forward-fill () for short price gaps, or drop rows where critical target labels are missing.

• Time alignment: Ensure uniform timestamps across multi-asset books and remove duplicate index entries. [1, 2]  

2. Normalize and Scale Numerical Features

• Fit strictly on train split: Prevent data leakage by computing scalers only on training sets.

• Choose method:

	• Use Standardization (Z-score: mean 0, variance 1) for linear models or neural networks.
	• Use Min-Max scaling or Robust scaling (if heavy outlier spikes exist in order-book size/volume).

• Rolling scaling: For streaming data, apply rolling z-scores over a fixed window (e.g., 20-period mean/std) to handle non-stationary price/volume distributions.

3. Engineer Volatility & Liquidity Features

• Volatility metrics:

	• Realized volatility (standard deviation of log returns over N periods, e.g., N=5, 20).
	• True Range (TR) or Average True Range (ATR).

• Liquidity metrics:

	• Bid-ask spread percentage.
	• Amihud illiquidity measure (absolute return / dollar volume).
	• Rolling average trade size, order-book depth imbalance, or turnover ratio.
